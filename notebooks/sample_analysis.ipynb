# Single-sample graph analysis

This is the only notebook for the single-response case-study workflow. Feature construction is implemented in `graph_features.py`; control matching, run analysis and plotting are implemented in `sample_analysis.py`.

```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == 'notebooks':
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT))

from sample_analysis import SampleAnalysis

DATA_ROOT = Path('/share/home/tm902089733300000/a903202310/lys/data/RAGTruth/model_traces/llama31_8b/test')
OUTPUT_ROOT = REPO_ROOT / 'outputs' / 'sample_analysis'
GENERATOR_MODEL = None
ERROR_SAMPLE_ID = '10071'
ERROR_INDEX = 0
PRE_WINDOW = 10
POST_WINDOW = 10
LOCAL_RADIUS = 18
MAX_CONTROLS = 32
```

```python
analysis = SampleAnalysis(
    DATA_ROOT,
    output_root=OUTPUT_ROOT,
    generator_model=GENERATOR_MODEL,
    random_state=0,
)
analysis.provenance()
```

```python
sample_id = str(ERROR_SAMPLE_ID) if ERROR_SAMPLE_ID is not None else analysis.error_sample_ids[ERROR_INDEX]
sample = analysis.dataset[sample_id]
print('sample_id:', sample_id)
print('observer_model:', sample.observer_model)
print('generator_model:', sample.generator_model)
print('positive_runs:', analysis.labels.positive_runs(sample_id))
analysis.runs(sample_id, pre_window=PRE_WINDOW, post_window=POST_WINDOW)
```

```python
result = analysis.visualize(
    sample_id,
    pre_window=PRE_WINDOW,
    post_window=POST_WINDOW,
    local_radius=LOCAL_RADIUS,
    max_controls=MAX_CONTROLS,
)
result['metadata']
```
